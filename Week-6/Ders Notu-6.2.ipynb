```python
#kütüphanelerin çağırılması

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
# Verinin çekilmesi

data = pd.read_excel("Adidas_US_Sales_Datasets.xlsx")
data
```

```python
data.head(10)
```

```python
data.tail(10)
```

```python
data.info()
```

```python
df = data.drop("Unnamed: 0", axis = 1)
df.head()

# Burada "Unnamed: 0" isimli bir sütun siliniyor.
# axis=1 → sütun demek.
# (Eğer axis=0 olsaydı, satır silerdi.)
```

```python
#Etiket isimlerini değiştirdik 
df = df.rename(columns = {"Unnamed: 1":"Retailer",
                            "Unnamed: 2": "Retailer ID",
                            "Unnamed: 3": "Invoice Date",
                            "Unnamed: 4": "Region",
                            "Unnamed: 5": "State",
                            "Unnamed: 6": "City",
                            "Unnamed: 7": "Product",
                            "Unnamed: 8": "Price per Unit",
                            "Unnamed: 9": "Units Sold",
                            "Unnamed: 10": "Total Sales",
                            "Unnamed: 11": "Operating Profit",
                            "Unnamed: 12": "Operating Margin",
                            "Unnamed: 13": "Sales Method" } )
```

```python
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()

# 1️⃣ df.isnull()
# DataFrame’deki her hücreyi kontrol eder. Eğer hücre boşsa (NaN, None), True yazar. Boş değilse False yazar.
# 2️⃣ .sum()
# Bu True–False değerlerini sütun bazında toplar. True = 1, False = 0 olduğu için, her sütunda kaç tane boş (eksik) değer olduğunu bulursun.
```

```python
df = df.drop([0,1,2,3], axis = 0) 
df.head(10)
```

```python
df.isnull().any()
```

```python
df = df.reset_index(drop=True)
df
```

```python
df.groupby("Retailer")[["Operating Margin"]].mean().reset_index()
```

```python
x = df.groupby(["Retailer","Region"])[["Units Sold"]].mean().reset_index()
x
```

```python
plt.figure(figsize=(15,5)) #Grafiğini boyutunu belirler
sns.barplot(x = "Region", y = "Units Sold", hue = "Retailer", data = x, palette = "viridis")
plt.xlabel("Region")
plt.ylabel("Units Sold")
plt.title("Bölgelere göre parakendecilerin satış miktarı")
plt.xticks(rotation=90);
```

```python
df.head()
```

```python
product_sales = df.groupby("Product")[["Units Sold"]].sum().reset_index()
product_sales
```

```python
plt.figure(figsize=(15,5)) #Grafiğini boyutunu belirler
sns.barplot(x = "Product", y = "Units Sold", data = product_sales, hue="Product", palette = "viridis",)
plt.xlabel("Ürün")
plt.ylabel("Toplam satış miktarı ")
plt.title("ürünlere göre toplam  satış miktarı")
plt.xticks(rotation=90);
```

```python
product_saledmetod = df.groupby(["Product","Sales Method"])[["Units Sold"]].sum().reset_index()
product_saledmetod
```

```python
px.bar(x= "Sales Method", 
       y="Units Sold", 
       color = "Product", 
       data_frame = product_saledmetod, 
       barmode = "group", 
       title = "Ürünlere göre satış miktarı")
```

```python
plt.figure(figsize=(25,7)) #Grafiğini boyutunu belirler
sns.barplot(x = "Sales Method", y = "Units Sold", hue = "Product", data = product_saledmetod, palette = "viridis")
plt.xlabel("Sales Method")
plt.ylabel("Units Sold")
plt.title("Satış Kanallarına göre satış miktarı")
plt.xticks(rotation=90);
```
